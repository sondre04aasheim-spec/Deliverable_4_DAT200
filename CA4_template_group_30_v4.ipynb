### Data description

The dataset contains information about the customers of a mobile phone operator. Each row is one customer. The features are:

   - CustomerId – unique ID of the customer (not a feature for modelling)
   - State – US state where the customer lives (two-letter code)
   - AccountLength – how long the customer has had an account
   - AreaCode – telephone area code of the customer
   - InternationalPlan – whether the customer has an international calling plan (yes/no)
   - VoiceMailPlan – whether the customer has a voice mail plan (yes/no)
   - NumVoiceMailMessages – number of voice mail messages
   - DayMinutes, DayCalls, DayCharge – total minutes, number of calls and total charge for daytime calls
   - EveMinutes, EveCalls, EveCharge – the same for evening calls
   - NightMinutes, NightCalls, NightCharge – the same for night calls
   - IntlMinutes, IntlCalls, IntlCharge – the same for international calls
   - CustomerServiceCalls – number of calls the customer has made to customer service

Target variable: Churn
   - 0 = customer stays
   - 1 = customer leaves the company

```python
# Taken from chapter_6,_part_1
# Creation of a pipeline
pipe_lr = make_pipeline(StandardScaler(),                   # Scaling of input features
                        PCA(n_components=2),                # Compression into principal components
                        LogisticRegression(random_state=1)) # Fit Logistic Regression with standard parameters
pipe_lr
```

```python
# Train all parts of the pipeline using all .fit() and .transform() automatically
pipe_lr.fit(X_train, y_train)

# Predict through the .transforms() and .predict()
y_pred = pipe_lr.predict(X_test)
print('Test Accuracy: %.3f' % pipe_lr.score(X_test, y_test))
```

### Scikit-learn Pipeline 2

```python
# Taken from chapter_6,_part_1
# Remember C parameter in SVC (low = high tolerance for error in classifier)
pipe_svm = make_pipeline(StandardScaler(),          # Scaling of input features
                        PCA(n_components=5),        # Compression into principal components
                        SVC(C=1, random_state=1))   # Fit Support Vector Classifier with standard parameters
pipe_svm
```

```python
# [X_train, X_test, y_train, y_test]
# Train all parts of the pipeline using all .fit() and .transform() automatically
pipe_svm.fit(X_train, y_train)

# Predict through the .transforms() and .predict()
y_pred = pipe_svm.predict(X_test)
print('Test Accuracy: %.3f' % pipe_svm.score(X_test, y_test))
```

### Cross-validation

```python
# Taken from chapter_7,_part_1

clf1 = LogisticRegression(solver='lbfgs', penalty='l2',
                          C=0.001, random_state=1)

clf2 = DecisionTreeClassifier(max_depth=1,
                              criterion='entropy',
                              random_state=0)

clf3 = KNeighborsClassifier(n_neighbors=1)
```

```python
# Pipeline for only logistic 
logistic = make_pipeline(StandardScaler(), clf1)

# Pipeline for only KNN 
knn = make_pipeline(StandardScaler(), clf3)

clf_labels = ["pipe_lr", "pipe_svm", "Logistic regression", "Decision tree", "KNN"]
```

```python
print('10-fold cross validation:\n')

# To see how well the individual (and earlier pipelines) algorythms can predict
for clf, label in zip([pipe_lr, pipe_svm, logistic, clf2, knn], clf_labels):
    scores = cross_val_score(clf, X_train, y_train,
                             cv=10, scoring='roc_auc')      # cv = how many runs each classifier runs
    print("ROC AUC: %0.2f (+/- %0.2f) [%s]"
          % (scores.mean(), scores.std(), label))
```

Hard voting (majority voting)

```python
mv_clf_hard = VotingClassifier(
    estimators=[
        ('pipe_lr', pipe_lr),
        ('pipe_svm', pipe_svm),
        ('logistic', logistic),
        ('Decision tree', clf2),
        ('knn', knn)
    ],
    voting='hard'
)
```

```python
clf_labels_mv = clf_labels + ['Voting (hard)']
all_clf = [pipe_lr, pipe_svm, logistic, clf2, knn, mv_clf_hard]


for clf, label in zip(all_clf, clf_labels_mv):
    scores = cross_val_score(
        estimator=clf,
        X=X_train,
        y=y_train,
        cv=10,
        scoring='accuracy'
    )

    print("Accuracy: %0.2f (+/- %0.2f) [%s]"
          % (scores.mean(), scores.std(), label))
```

Soft voting (probability-based)

```python
# Since soft voting is probabilty based, pipe_svm needs to have probabilty true for it to work here
# Hence the sudden redefinition of pipe_svm

pipe_svm = make_pipeline(
    StandardScaler(),
    SVC(
        kernel='rbf',
        C=1.0,
        gamma='scale',
        probability=True,
        random_state=1
    )
)
```

```python
mv_clf_soft = VotingClassifier(
    estimators=[
        ('pipe_lr', pipe_lr),
        ('pipe_svm', pipe_svm),
        ('logistic', logistic),
        ('Decision tree', clf2),
        ('knn', knn)
    ],
    voting='soft'
)
```

```python
clf_labels_soft = clf_labels + ['Voting (soft)']
all_clf_soft = [pipe_lr, pipe_svm, logistic, clf2, knn, mv_clf_soft]

for clf, label in zip(all_clf_soft, clf_labels_soft):
    scores = cross_val_score(clf, X_train, y_train,
                             cv=10, scoring='roc_auc')
    print("ROC AUC: %0.2f (+/- %0.2f) [%s]"
          % (scores.mean(), scores.std(), label))
```
